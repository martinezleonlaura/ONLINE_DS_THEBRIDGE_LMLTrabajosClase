* Aplícalo para encontrar los duplicados de "un", "es" y "binario" de:  

>```ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]```


***EJERCICIO 2: FUNCIONA CORRECTAMENTE***

`````python
#Simplificando tengo que recorrer un lista, una lista de veces y crear otra con la solución

ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]
encontrar_duplicados = ["un", "es", "binario"]
indice_elem_duplicados = []

for duplicado in encontrar_duplicados:
    for i in range(len(ejer_3)):
        if ejer_3[i] == duplicado:
            indice_elem_duplicados.append(i)

print("Índice de los duplicados de ", encontrar_duplicados, ":")
print(sorted(indice_elem_duplicados))
`````

## Ejercicio 12

***EJERCICIO 12: FUNCIONA CORRECTAMENTE***

`````python
#Creo in input solicitando el email. Es un str.
email = input("Ingrese su dirección de email:")

#Defino una función función que me diga si el email es válido o no. Es decir si la str email contiene @ o no
def validar_email(email):
    if "@" in email:
        return True
    else:
        return False

#Ahora compruebo el email
if validar_email(email) == True:
    print("Gracias por facilitar su email:",email)
else:
    print("El email", email, "no es válido")
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


***EJERCICIO 13: FUNCIONA CORRECTAMENTE***

`````python
#Creo in input solicitando el DNI. Es un str.
dni_usuario = input("Introduzca su DNI con la letra en mayúsculas:")

#Creo un función para validar el "comprobar_dni"
def comprobar_dni(dni_usuario):

    digitos_numericos = dni_usuario [0:-1] #recuerda que se queda un valor anterior al último 
    letra = dni_usuario [-1]

    if len(digitos_numericos) < 7 or len(digitos_numericos) >8:
        return False

    else:
        DNI = int(dni_usuario [0:-1])
        palabra='TRWAGMYFPDXBNJZSQVHLCKE'

        if letra == palabra[DNI%23]:
            return True
        else:
            return False

print("Comprobación del DNI ", dni_usuario, ":", comprobar_dni(dni_usuario))
`````
