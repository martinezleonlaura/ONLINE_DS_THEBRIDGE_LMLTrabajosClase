## Ejercicio 1
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
vble_a = 4 # Asignación de vbles
vble_b = "Text"
vble_c = 4.1

print (vble_a == vble_b)
print (vble_a != vble_c)
print (vble_a > vble_c)
print (vble_c <= vble_a)
print (vble_b != vble_c)
````

## Ejercicio 2
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
color_cielo = input ("¿De qué color es el cielo? ") # Asignación de 2 inputs
color_mar = input ("¿De qué color es el mar? ")
print ("El cielo y el mar son de color azul:", color_cielo == color_mar) # Comprabación de si esos inputs son iguales o no
````

## Ejercicio 3
Mismo programa que en 2, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
arco_iris = input ("¿De cuántos colores principales está compuesto el arco iris? ") # Asignación de 3 inputs
dias_semana = input ("¿Cuántos días tiene una semana? ")
numero = input ("Cuál es tu número de la suerte? ")

print ("Todos son iguales: ", arco_iris == dias_semana and dias_semana == numero) # Comprabación de si esos inputs son los 3 iguales o no

print ("Al menos dos son iguales: ", arco_iris == dias_semana or dias_semana == numero) # Comprabación de si al menos 2 de esos 3 inputs son iguales o no
````
