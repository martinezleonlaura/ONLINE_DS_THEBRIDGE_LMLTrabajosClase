## Ejercicio 1
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

1. Calcula el máximo de la lista: [4, 6, 8, -1]
2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965
4. Valor absoluto de: -74

```python
lista_num1 = [4, 6, 8, -1]
print ("El número máximo de la lista es:", max(lista_num1))

print ("La suma de todos los elementos de la lista es:",sum (lista_num1))

numero_2 = 63.451256965
print ("El número 63.451256965 redondeado a 3 dígitos decimales es:", round(numero_2, 3))

numero_3 = -74
print ("El valor absoluto de -74 es:", abs (numero_3))
```

## Ejercicio 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

```python
str_quijote = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

print (str_quijote.upper()) # 1. Pásalo todo a mayúsculas

print (str_quijote.lower()) # 2. Pásalo todo a minúsculas

print (str_quijote.title()) # 3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula. Esta función no la encontré en los apuntes, la busqué en internet

lista_quijote = str_quijote.split() # 4. Crea una lista dividiéndolo por sus espacios
print (lista_quijote)

print (str_quijote.replace (",",";")) # 5. Sustituye las comas , por puntos y comas ;

print (str_quijote.replace("a","")) # 6. Elimina las a minúsculas
```

## Ejercicio 3
1. Crea una lista con 3 elementos numéricos
2. Añade un cuarto elemento
3. Calcula la suma de todos
4. Elimina el segundo elemento de la lista
5. Añade otro elemento en la posicion 3 de la lista
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.
7. Ordena la lista de menor a mayor.
8. BONUS: Ordena la lista de mayor a menor.  
(*PISTA: Para hacer 8, tendrás que buscar en la información del método...*)  
  
  
NOTA: Pinta por pantalla el resultado de cada paso.

```python
lista = [1, 2, 3] # 1. Crea una lista con 3 elementos numéricos
print (lista)

lista.append (4) # 2. Añade un cuarto elemento
print (lista)

print ("La suma de todos los elementos numéricos de mi lista es:", sum(lista)) # 3. Calcula la suma de todos

lista.pop(1) # 4. Elimina el segundo elemento de la lista. OJO!! la primera posición es cero, por lo que la segunda será 1
print (lista)

lista.insert (3, 5) # 5. Añade otro elemento en la posicion 3 de la lista
print (lista)

lista_2 = [6, 7, 8, 9] # 6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.
concat = lista_2 + lista
print (concat)

concat.sort() # 7. Ordena la lista de menor a mayor.
print (concat)

concat.sort(reverse=True) # 8. BONUS: Ordena la lista de mayor a menor.
print (concat)
```
