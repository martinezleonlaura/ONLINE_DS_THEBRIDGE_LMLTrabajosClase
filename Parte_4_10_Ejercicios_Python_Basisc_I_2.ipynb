## Ejercicio 1

1. Crea dos variables numericas: un `int` y un `float`
2. Comprueba sus tipos
3. Sumalas en otra nueva
4. ¿De qué tipo es la nueva variable?
5. Elimina las dos primeras variables creadas

```python
#Crea dos variables numericas: un int y un float
entero = 5
real = 10.5

#Comprueba sus tipos
print(entero, type (entero))
print(real, type (real))

#Sumalas en otra nueva
suma = entero + real

#¿De qué tipo es la nueva variable?
print (suma, type (suma))

#Elimina las dos primeras variables creadas
del(entero)
del(real)
```

## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
grados = float(input ("Introduce el ángulo en grados que quieres pasar a radianes ")) # Lo he pasado a real para que pueda hacer la multiplicación por pi
import math # Para poder aplicar el número pi tenoge que importar primero
radianes = round(grados * (math.pi / 180),2)
print (grados, "grados son", radianes, "radianes")
```

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

```python
base_1 = float(input("Introduce la base del paralelgramo en metros "))
altura_1 = float (input("Introduce la altura del paralelogramo en metros "))
area_1 = round(base_1 * altura_1, 2)
print ("El área del paralelogramo es de ", area_1, "m2")
```
